#Group by and aggregating

```python
import pandas as pd
```

```python
df = pd.read_csv('/Users/apple/Projects/python/Anaconda Lesson/data of pandas reading exercise /Flavors.csv')
```

```python
df
```

```python
group_by_frame = df.groupby('Base Flavor')
```

```python
group_by_frame
```

```python
group_by_frame.mean(numeric_only=True)
```

```python
group_by_frame[['Flavor Rating', 'Texture Rating', 'Total Rating']].mean()
```

```python
group_by_frame.mean(numeric_only=True)
```

```python
count_by_flavor = df.groupby('Base Flavor').count()
count_by_flavor
```

```python
group_by_frame.count()
```

```python
group_by_frame.min()
```

```python
group_by_frame.max()
```

```python
group_by_frame.sum(['Flavor Rating', 'Texture Rating', 'Total Rating'])
```

```python
group_by_frame.agg({'Flavor Rating': ['mean','max','min','count','sum']})
```

```python
group_by_frame.agg({'Flavor Rating': ['mean','max','min','count','sum'],'Texture Rating': ['mean','max','min','count','sum']})
```
