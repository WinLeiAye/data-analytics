```python
daily_pints = {1,2,3}
```

```python
type(daily_pints)
```

```python
daily_pints_log = {1,2,31,2,3,4,1,2,5,6,3,2,}
```

```python
daily_pints_log 
```

```python
wifes_daily_pints_log = {1,3,5,7,3,24,5,7,3,2,0}
```

```python
print (daily_pints_log | wifes_daily_pints_log)
#combine all value but all uniqe between two sets
```

```python
print (daily_pints_log & wifes_daily_pints_log)
##show both set 
```

```python
print (daily_pints_log - wifes_daily_pints_log)
#what does not match
```

```python
print (wifes_daily_pints_log - daily_pints_log)
#substract and show remaining values
```

```python
print (wifes_daily_pints_log ^ daily_pints_log)
#value either one or other not in both. 
```
