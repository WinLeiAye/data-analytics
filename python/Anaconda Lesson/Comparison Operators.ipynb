###Logical Operator
and - returns True if both statements are true
or  - returns True if one of the statements is true
not - reverse the result, returns False if the result is true 

```python
10 > 50
```

```python
(10 > 50) and (50 > 10)
```

```python
(10 > 50) or (50 > 10)
```

```python
(70 > 50) and (50 > 10)
```

```python
('Vanilla' > 'Chocolate') and (50 > 10)
```

```python
not  (50 > 10)
```

```python
###Membership Operators
in _ Returns True if a sequence with the specified value is present in the object
not in_ Returns True if a sequence with the specified value is not present in the object
```

```python
ice_cream = ' I love chocolate ice cream'
'love' in ice_cream
```

```python
scoops = [1,2,3,4,5]
2 in scoops
```

```python
scoops = [1,2,3,4,5]
6 in scoops
```

```python
scoops = [1,2,3,4,5]
6 not in scoops
```

```python
scoops = [1,2,3,4,5]
wanted_scoops = 8

wanted_scoops in scoops
```

```python
wanted_scoops not in scoops
```
