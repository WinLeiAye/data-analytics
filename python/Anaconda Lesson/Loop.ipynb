##Nested for loops

```python
flavours = ['vanilla', 'chocolate', 'cookie dough']
toppings = ['hot fudge', 'oreos', 'marshmallow' ]
```

```python
for one in flavours :
    for two in toppings:
        print (one, 'topped with', two)
```
