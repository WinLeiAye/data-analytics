```python
##Merge,Join, and Concatenate
```

```python
import pandas as pd
```

```python
df1 = pd.read_csv('/Users/apple/Projects/python/Anaconda Lesson/data of pandas reading exercise /LOTR.csv')
```

```python
df1
```

```python
df2 = pd.read_csv('/Users/apple/Projects/python/Anaconda Lesson/data of pandas reading exercise /LOTR 2.csv')
```

```python
df2
```

```python
df1.merge(df2) #inner join
```

```python
df1.merge(df2, how = 'inner')
```

```python
df1.merge(df2, how = 'inner', on = 'FellowshipID')
```

```python
df1.merge(df2, how = 'inner', on = ['FellowshipID','FirstName'] )
```

```python
df1.merge(df2, how = 'outer')
```

```python
df1.merge(df2, how = 'left')
```

```python
df1.merge(df2, how = 'right')
```

```python
df2.merge(df1, how = 'left')
```

```python
df2.merge(df1, how = 'outer')
```

```python
df1.merge(df2, how = 'cross')
```

```python
df1.join(df2, on = 'FellowshipID', how = 'outer', lsuffix = '_left', rsuffix = '_right')
```

```python
df4 = df1.set_index('FellowshipID').join (df2.set_index('FellowshipID'),lsuffix = '_left', rsuffix = '_right', how = 'inner')
```

```python
df4
```

```python
df5 = df1.set_index('FellowshipID').join (df2.set_index('FellowshipID'),lsuffix = '_left', rsuffix = '_right', how = 'outer')
```

```python
df5
```

##Concatenate

```python
pd.concat([df1,df2])
pd.concat([df1,df2], join = 'inner')
pd.concat([df1,df2], join = 'outer', axis = 1)
```
